Imports & Seeding

```python
import random as r
r.seed(1)
```

# Knapsack Problem

```python
weight_limit = 100
items = [
    [11, 53], #Laptop
    [54, 51], #Mini Fridge
    [21, 46], #Book
    [5, 37], #Playing cards
    [7, 22], #Sun hat
    [2, 10], #Calculator
    [33, 37], #Swimsuit
    [49, 45], #Onesie
    [12, 15], #Beer
    [20, 27] #Wine
]

def reward(solution, item_list = items):
    assert(len(solution) == len(item_list))
    utility = 0
    weight = 0
    for i in range(len(item_list)):
        if solution[i] == 1:
            item = items[i]
            weight += item[0]
            utility += item[1]
    return(0 if weight > weight_limit else utility)
```

# Genetic Algorithm

```python
genetic_fitness = reward
num_individuals = 10
num_iterations = 1000

#For the initial population
def generate_individual():
    genes = [r.choice([0, 1]) for i in range(len(items))]
    return([genes, genetic_fitness(genes)])

#If we mutate we swap 1 -> 0 or 0 -> 1
def mutate_genome(genome, mutation_rate = 0.1):
    output_genome = [g for g in genome]
    for i in range(len(genome)):
        if r.uniform(0, 1) < mutation_rate:
            output_genome[i] = 1 - output_genome[i]
    return(output_genome)

#Note that we may pick the same parent twice. This is not realistic but not a problem for the algorithm.
def breed(population):
    to_reproduce_1 = r.choice(population)
    to_reproduce_2 = r.choice(population)
    genome = [to_reproduce_1[i] if r.uniform(0, 1) < 0.5 else to_reproduce_2[i] for i in range(len(to_reproduce_1))]
    genome = mutate_genome(genome[0])
    offspring = [genome, genetic_fitness(genome)]
    stopping_condition = False
    i = 0
    while not stopping_condition:
        if offspring[1] > population[i][1]:
            population[i] = offspring
            stopping_condition = True
        if i == len(items) - 1:
            stopping_condition = True
        i += 1
    return(population)

population = [generate_individual() for n in range(num_individuals)]
for n in range(num_iterations):
    population = breed(population)
print(f"solution: {population[0]}")
```
