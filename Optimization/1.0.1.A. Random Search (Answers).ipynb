$R(v)$ gives us the resistance per second. If we travel a constant distance D then we travel for $\frac{D}{v}$ seconds. Minimizing the resistance is therefore the same as minimizing $\frac{D}{v} \cdot R(v)$. $D$ is constant so we can just minimize:

$L(x) = \frac{a}{x} + b + cx$

```python
import random as r
r.seed(1)

def generate_hypothesis():
    return(r.uniform(0, 100))
    
def loss(x, a = 200, b = 0.01, c = 0.3):
    return(a / x + b + c * x)

num_iterations = 10 ** 4
x_star = generate_hypothesis()
y_star = loss(x_star)
for i in range(num_iterations):
    x_new = generate_hypothesis()
    y_new = loss(x_new)
    if y_new < y_star:
        y_star = y_new
        x_star = x_new
print(f"Random search produces veolcity: {x_star:.2f}m/s")
```

Which works out to around 58mph or 93kph.
