We will minimize $L(x)$ which is a many-dimensional function, using gradient descent.

For $L(x)$ we pick the sphere function with n = 5: https://www.sfu.ca/~ssurjano/spheref.html

Which is defined by:

$f(x) = \sum_{i = 1}^{5} {x_i} ^ {2}$

```python
#Imports and seeding
import random as r
r.seed(1)
```

# Loss Function

```python
#The function we minimize
def loss(x):
    return(sum(xi**2 for xi in x))

num_dimensions = 5 #Length of x
x_range = [-10, 10] #Min/max values of each xi in x
```

# Gradient Descent

```python
def calculate_derivative_component(x, loss, dimension_num, dx = 10 ** -3):
    x_copy = [xi for xi in x]
    x_copy[dimension_num] += dx
    return((loss(x_copy) - loss(x)) / dx)

def delta_vector(x, loss, dx = 10 ** -5, learn_rate = 10 ** -2):
    return([-learn_rate * calculate_derivative_component(x, loss, i, dx) for i in range(len(x))])
    
def gradient_descent(loss, dx = 10 ** -5, learn_rate = 10 ** -2, num_iterations = 1000):
    x = [r.uniform(x_range[0], x_range[1]) for n in range(num_dimensions)]
    for i in range(num_iterations):
        delta = delta_vector(x, loss, dx = dx, learn_rate = learn_rate)
        x = [x[i] + delta[i] for i in range(len(x))]
    return(x)

val = gradient_descent(loss)
print(val)
```
